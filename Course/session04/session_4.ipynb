Loops: Doing Things Repeatedly

Computers are great at doing boring tasks again and again — automatically.

While

```python
counter = 0
while counter < 10: # 0 to 99 = 100
  print('salavat')
  counter = counter + 1
```

```python
count = 0
while count < 5:
    print("Hello!")
    count += 1
```

```python
while True:
    print("This never ends!")
```

Use when you know how many times you want to repeat

```python
for i in range(5):
    print("Loop number:", i)
```

```python
range(5)        # 0 → 4
range(1, 6)     # 1 → 5
range(10, 0, -1)  # Countdown: 10 → 1
```

 Repeat After Me

```python
word = input("Give me a word: ")
times = int(input("How many times? "))

for i in range(times):
    print(f"{i+1}. {word}")
```

Secure Login Retry

```python
password = "python123"
attempt = ""

while attempt != password:
    attempt = input("Enter password: ")

print("✅ Access granted!")
```
