## مراحل تبدیل
<div dir="rtl" align="right">
به‌صورت خلاصه، مراحلی که یک کد ++C برای کامپایل شدن، یا همان تبدیل شدن به زبان قابل‌فهم کامپیوتر طی می‌کند به شکل زیر است:


<div dir="rtl" align="right">
در ابتدا، کد نوشته‌شده‌ی ما به پردازنده‌ی مدنظر داده می‌شود تا پردازش اولیه مثل بررسی کردن `<include>`‌ها و مسیر‌های مختلف را انجام دهد. اکثر خطا‌های کد در این مرحله پیدا خواهند شد.
<div dir="rtl" align="right">
در مرحله‌ی بعدی، کد ما به یک زبان سطح پایین مثل اسمبلی تبدیل خواهد شد تا در سطح سخت‌افزار قابل اجرا باشد. اسمبلی در واقع زبانی است که فهم آن برای ماشین ساده‌تر است.
<div dir="rtl" align="right">
در ادامه‌، کد اسمبلی به همراه تمام کتابخانه‌ها و یا فایل‌های کدی که در برنامه‌ی ما وجود دارد، تبدیل به یک فایل قابل‌اجرا (Executable File) شده و در نهایت بر روی سیستم ما اجرا می‌شود.



</div><div dir="rtl" align="right">
اما در پایتون از این ساختار استفاده نمی‌شود و به جای آن، کد شما تفسیر خواهد شد. اما تفسیر به چه معناست؟



<div dir="rtl" align="right">
کد پایتونی ما مسیر متفاوتی را طی خواهد کرد. به جای این‌که تمام کد‌ ما به‌یک‌باره تبدیل به کد اسمبلی و فایل قابل‌اجرا شود، این اتفاق مثل مسیر بالا به صورت خط‌به‌خط خواهد افتاد.
<div dir="rtl" align="right">
در ابتدا، کد ما به کوچک‌ترین واحد نشانه‌ای یا همان توکن‌ها تبدیل می‌شود. توکن در واقع کلمات کلیدی مثل if، for و ... هستند که در کد ما وجود دارد.
<div dir="rtl" align="right">
در مرحله‌ی بعدی توکن‌های کد ما بررسی می‌شوند تا خطاهای تایپی یا معنایی آن مشخص شود.
<div dir="rtl" align="right">
در ادامه‌، کل کد ما تبدیل به یک درختی انتزاعی سینتکسی (Abstract Syntax Tree) خواهد شد. به زبان ساده، این درخت انتزاعی سینتکسی در واقع تمام کد‌ شماست که ماشین راحت‌تر متوجه آن می‌شود.



<div dir="rtl" align="right">
در نهایت کد ما به یک مفسر زبان پایتون که وظیفه‌ی تبدیل کد به زبان ماشین را دارد، داده می‌شود و این مفسر، کد درخت شما را خط‌به‌خط پیمایش می‌کند و دستورات مربوطه‌ در آن اجرا می‌شوند. برای مثال ایجاد متغیر‌ها، ساختار‌های کنترلی و صدا کردن یک تابع در این مرحله انجام می‌شود.
<div dir="rtl" align="right">
بعد از این مرحله‌، نوبت به تولید ‌Bytecode می‌رسد. بایت‌کد در واقع چیزی مشابه همان فایل‌های اسمبلی است، اما یک مرحله‌ نزدیک‌تر به زبان ماشین یا همان صفر و یک.
<div dir="rtl" align="right">
در نهایت این بایت‌کد توسط مفسر ما اجرا شده و نتیجه را در خروجی برای ما نمایش خواهد داد.

## تمرین عملیات بیتی

### سوالات نظری
1. تفاوت بین عملگر `&` و عملگر `|` را با ذکر مثال توضیح دهید.
2. چرا نتیجه عملگر `~9` برابر با `-10` می‌شود؟ (مربوط به نمایش اعداد در مبنای دو)
3. اگر `x = 12` باشد، خروجی `x >> 2` چه مقدار خواهد بود و چرا؟

### سوالات عملی
1. برنامه‌ای بنویسید که:
   - عددی از کاربر بگیرد
   - آن را دو بار به چپ شیفت دهد
   - سپس نتیجه را با عملگر `&` و عدد 15 مقایسه کند
   - خروجی نهایی را نمایش دهد

2. با استفاده از عملگرهای بیتی، تابعی بنویسید که:
   - بررسی کند یک عدد زوج است یا فرد (راهنمایی: از عملگر `&` استفاده کنید)
   - بدون استفاده از عملگر `/` یا `*` عدد را در ۴ ضرب کند (راهنمایی: از شیفت استفاده کنید)
   - بدون استفاده از عملگر `/` یا `%` باقیمانده تقسیم بر ۸ را محاسبه کند

### تمرین  
- برنامه‌ای بنویسید که دو عدد از کاربر بگیرد و:
  - نتیجه `AND`، `OR` و `XOR` بیتی آنها را نمایش دهد
  - هر دو عدد را به اندازه عدد دوم به چپ شیفت دهد و نتایج را مقایسه کند
  - معکوس بیتی هر دو عدد را محاسبه و نمایش دهد

## عملگرهای منطقی

| عملگر | مثال | توضیحات |
|:-----:|:----:|:-------:|

```python
x = {"age": 19, "name": "maryam"}
y = {"name": "maryam", "age": 19}
z = x
w1 = 1
w2 = 1
print(x is y)  # Output: False
print(y is not z)  # Output: True
print(x == y)  # Output: True
print(w1 is w2)  # Output:True
```

## عملگرهای بیتی
<div dir="rtl" align="center">
از عملگرهای بیتی برای تغییر تک‌تک بیت‌های اعداد صحیح استفاده می‌شود.

| عملگر | نام | توضیحات |
|:-----:|:---:|:-------:|
| `&` | AND | اگر هر دو عملوند 1 باشند نتیجه 1 خواهد بود و در غیر این صورت 0 است. |
| `\|` | OR | اگر هر دو عملوند 0 باشند نتیجه 0 خواهد بود و در غیر این صورت 1 است. |
| `^` | XOR | اگر حداکثر یکی از عملوندها 1 باشد نتیجه 1 خواهد بود و در غیر این صورت 0 است. |
| `~` | NOT | همه‌ی بیت‌ها را معکوس می‌کند. |
| `<<` | Left Shift | مقدار عملوند سمت چپ به اندازه بیت‌های مشخص‌شده به چپ منتقل می‌شود (معادل ضرب در ۲). |
| `>>` | Right Shift | مقدار عملوند سمت چپ به اندازه بیت‌های مشخص‌شده به راست منتقل می‌شود (معادل تقسیم بر ۲). |


```python
a = 13 == 13
b = 12 == 12
print(b, a)
c = a & b
c
```

```python
print(9 << 2)  # Output: 36
# 9 = 0000000000001001  -> 36 = 0000000000100100
```

## Why Use f-strings?
Cleaner, readable, and flexible string formatting

```python
name = input("Name: ")
age = int(input("Age: "))
print(f"{name} is {age} years old.")
```

```python
print("Hello " + name + ", you are " + str(age) + " years old.")     # Harder
```

## Type Conversion

```python
a = 3.14  # float
a = int(a)
b = "3"  # string
b = int(b)
print(a)  # Output: 3
print(b)  # Output: 3
```

```python
a = 1
a = float(a)
b = "3.14"
b = float(b)
print(a)  # Output: 1.0
print(b)  # Output: 3.14
```

```python
number = 42
str_number = str(number)
print(str_number)  # Output: '42'
```
