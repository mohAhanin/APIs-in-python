## Type Conversion

````python
a = 3.14  # float
a = int(a)
b = "3"  # string
b = int(b)
print(a)  # Output: 3
print(b)  # Output: 3
````

````python
a = 1
a = float(a)
b = "3.14"
b = float(b)
print(a)  # Output: 1.0
print(b)  # Output: 3.14
````

````python
number = 42
str_number = str(number)
print(str_number)  # Output: '42'
````

What is the random library?

The random library is a built-in module in Python that lets you generate random numbers or make random choices. You use it when your program needs to simulate chance or unpredictability — like in games, simulations, or quizzes.

Why use it?

1. To simulate dice rolls

2. To randomly choose a question from a quiz

3. To create unpredictable behavior in games (like a mystery number)

4. To shuffle items like a deck of cards



````python
import random
````

````python
import random

number = random.randint(1, 10)
print("I'm thinking of a number between 1 and 10...")
guess = int(input("What’s your guess? "))

if guess == number:
    print("🎉 You got it!")
else:
    print(f"Sorry, it was {number}")
````

It’s a simple Python game you can build with students where:

1. The computer chooses a random number between 1 and 10

2. The user has 3 chances to guess it

3. After each guess, the program tells the user if the guess is too high or too low

4. If they guess correctly, the program congratulates them
