## Introducing input()

The input() function lets the user type something, and the program stores it.

Imagine you're programming a robot to ask questions and then respond based on the answers.

```python
name = input("What is your name? ")
print("Hello, " + name + "!")
```

```python
color = input("What is your favorite color? ")
print("Wow, I like " + color + " too!")
```

```python
age = input("How old are you? ")
print(age + 1)  # 🚫 Error!
```

```python
age = int(input("How old are you? "))
print("Next year, you'll be", age + 1)
```
