### 📡 Step 3: Basic API Request
Let’s make a simple request to OpenRouter.

```python
import requests

API_KEY = "your-api-key-here"  # 🔑 Replace with your key!
API_URL = "https://openrouter.ai/api/v1/chat/completions"

headers = {
    "Authorization": f"Bearer {API_KEY}",
    "Content-Type": "application/json"
}

data = {
    "model": "deepseek/deepseek-chat:free",  # 🏷️ Choose any model
    "messages": [
        {"role": "user", "content": "Tell me a joke about cats!"}
    ]
}

response = requests.post(API_URL, headers=headers, json=data)

if response.status_code == 200:
    answer = response.json()
    print("🤖 AI Response:", answer["choices"][0]["message"]["content"])
else:
    print("❌ Error:", response.text)
```

### What’s happening?
We send a POST request to OpenRouter.

We specify the AI model (gpt-3.5-turbo).

We pass a user message (Tell me a joke about cats!).

The AI responds, and we print it.

### 📌 Step 5: Error Handling
Sometimes the API fails. Handle errors like this:

```python
response = requests.post(API_URL, headers=headers, json=data)

if response.status_code == 200:
    print("✅ Success:", response.json())
elif response.status_code == 401:
    print("🔒 Error: Invalid API Key!")
elif response.status_code == 429:
    print("🚦 Error: Too many requests! Slow down.")
else:
    print("❌ Unknown Error:", response.text)
```

### Recap

```python
# Our secret password to talk to the robot
API_KEY = ''

# The robot's address on the internet
API_URL = 'https://openrouter.ai/api/v1/chat/completions'

# How we introduce ourselves to the robot
headers = {
    'Authorization': f'Bearer {API_KEY}',  # Showing our secret password
    'Content-Type': 'application/json'     # Telling it we're sending text
}

your_question = 'what do you know about Shiraz?'

# Prepare what to send
message_to_send = {
    "model": "deepseek/deepseek-chat:free",
    "messages": [
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": your_question}
    ]
}

# Send the question and get the answer
response = requests.post(API_URL, headers=headers, json=message_to_send)

# Show the answer
if response.status_code == 200:
    answer = response.json()
    print("🤖 Answer:", answer["choices"][0]["message"]["content"])
else:
    print("❌ Error:", response.status_code, response.text)
```
